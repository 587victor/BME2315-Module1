* Prevalence & incidence

Alzheimer is a type of brain disease that is a common cause of dementia and brings memory loss and damages thinking skills. According to the National Institute of Aging, more than 6 million Americans aged 65 and older may have Alzheimer’s, and age is the biggest known risk factor for Alzheimer’s. Rates of Alzheimer disease are expected to increase as the population gets older. An estimated 7.4 million Americans aged 65 and older are living with Alzheimer's disease which around 11% of them aged 65 and older have Alzheimer's. Lifetime risk at age 45 is higher for women (1 in 5) than men (1 in 10), and nearly two-thirds of those affected are women. About 200,000 Americans under 65 have dementia that started  early. By 2060, about 13.8 million older Americans are expected to have Alzheimer's.

* Risk factors (genetic, lifestyle) 
Alzheimer’s disease can be different because most people with the disease are 65 and older, and according to the Alzheimer's Association, the risk of Alzheimer’s doubles every five years after adults with 65 get older. Another huge factor for Alzheimer's disease is called APOE, which stands for apolipoprotein E. Everyone inherits one gene from each parent, but people who inherit the ε4 version have a harder time clearing a sticky protein called amyloid from the brain, which increases their chances of getting Alzheimer's disease, and it slowly builds up over years and damages brain cells, which can lead to memory loss later in life. Having a parent or sibling with Alzheimer's increases the risk about 2 to 3 times. Additionally, people who have heart problems are at risk of getting Alzheimer's disease such as; having high blood pressure, diabetes, high cholesterol, and obesity all raise the risk.


* Societal determinants: 
Societal determinants can play an important role in developing Alzheimer’s disease and people’s ability to receive better care. Societal determinants  such as education, income, access to healthcare, neighborhood environment, and social support can all play a role. According to the National Institutes of Health, people who have higher levels of educational background have a lower risk of dementia and better cognitive function later in life. One possible reason is that education challenges and keeps the brain and builds cognitive reserve, which may allow the brain to continue functioning even when age-related changes or damage happen. Another Societal factor that causes Alzheimer’s at an early age is income because people with lower income tend to have less access to healthy food or healthcare for early diagnosis, and living in unsafe neighborhoods and environment that can affect the brain.

* Symptoms
    There are many symptoms of Alzheimer's disease, that includes Memory loss that disrupts daily life, especially forgetting recently learned information; trouble planning or solving problems, such as following a recipe, managing bills, or keeping track of monthly tasks; poor judgment; greater memory loss, including not recognizing family or friends, not being able to find your address or getting lost, needing help with basic daily tasks, having difficulty with speaking, swallowing, and walking. Another symptom of Alzheimer's is communication problems, including having difficulty following a conversation and losing track of time and date.
* Diagnosis
Doctors don’t have a single test to determine if someone has Alzheimer’s, they combine different tests such as: asking about memory and thinking changes, cognitive or screening test, MRI or CT imaging to see certain parts of brain’s shrinkage.Also, a  special PET scan can show the harmful proteins such as amyloid and tau that are linked to Alzheimer's and doctors often repeat tests months apart and if  Alzheimer's patients memory and thinking keeps getting worse over time, it may support the diagnosis .In the past, Alzheimer’s could only be confirmed after a person died by examining the brain. Today, doctors can diagnose Alzheimer’s while a person is still alive with much more certainty.

* Standard of care treatments (& reimbursement):

Scientists have found no cure for Alzheimer's disease, and none of the current treatments can stop or treat an Alzheimer's patient, but there are some treatments that can help to manage symptoms of this disease and slow down its progression.  According to the U.S. Food and Drug Administration, some medicines, such as donepezil, rivastigmine, and galantamine, help the human brain with memory and thinking, and Memantine is helpful in protecting brain cells from too much chemical activity. Other medicines, such as Leqembi and Kisunla, help remove harmful protein buildup in the brain, which helps with Alzheimer disease. Additionally, other treatments for Alzheimer’s disease can be exercising and controlling blood pressure, diabetes, and cholesterol that are the main causes of Alzheimer's.
